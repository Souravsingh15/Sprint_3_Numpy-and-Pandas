#1. Datetime


Datetime is used to work with both date and time information in Python and Pandas.

It allows us to store, compare, manipulate, and analyze dates and times.

Pandas provides functions such as `pd.to_datetime()` to convert values into datetime format.

## Demonstration

### Simple Example

Suppose a dataset contains the date when each customer placed an order.

We can convert these values into datetime format so that we can perform date-based operations.

### Business Example

A company can use datetime information to analyze daily sales and identify which days have the highest sales.

### AI/ML Example

Datetime information can be converted into useful features such as day, month, year, or weekday before training a machine learning model.

```python
import pandas as pd

dates = pd.Series(["2026-08-20", "2026-08-21", "2026-08-22"])

dates = pd.to_datetime(dates)

print(dates)
```

## Explanation

`pd.to_datetime()` converts the values into Pandas datetime objects.

After conversion, we can perform operations such as extracting the year, month, day, or weekday.

## Where It Is Used

Datetime is commonly used in sales analysis, time-series analysis, financial data, customer activity, and machine learning feature engineering.

#2. Timestamp


A Timestamp represents a specific point in time.

Pandas uses `pd.Timestamp()` to create timestamp objects.

It provides more detailed information than just a date because it can include the date, time, and other time-related information.

## Demonstration

### Simple Example

If we want to represent exactly when an event happened, such as a customer placing an order at 10:30 AM, we can use a Timestamp.

### Business Example

An online shopping company can store the exact time when an order was placed.

### AI/ML Example

Timestamp information can be useful for analyzing user activity and creating time-based features.

```python
import pandas as pd

timestamp = pd.Timestamp("2026-08-20 10:30:00")

print(timestamp)
```

## Explanation

`pd.Timestamp()` creates a Pandas Timestamp object representing a specific date and time.

## Where It Is Used

Timestamps are useful for transaction records, logs, user activity, sensor data, and time-series analysis.

#3. Date Parsing


Date parsing means converting date values stored as strings into a proper datetime format.

Datasets may contain dates in different formats such as:

- `20-08-2026`
- `2026/08/20`
- `August 20, 2026`

Pandas can convert these values into datetime objects using `pd.to_datetime()`.

## Demonstration
### Simple Example

Suppose a dataset contains dates as text.

Before performing date calculations, we should convert those text values into datetime format.

### Business Example

A company may receive customer order dates as strings from a CSV file. Date parsing allows the company to analyze orders by month or year.

### AI/ML Example

Before creating time-based features, date columns need to be converted into a proper datetime format.

```python
import pandas as pd

data = pd.Series(["2026-08-20", "2026-09-15", "2026-10-10"])

parsed_dates = pd.to_datetime(data)

print(parsed_dates)
```

## Explanation

`pd.to_datetime()` parses the string values and converts them into datetime values.

This allows us to perform calculations and extract date features.

## Where It Is Used

Date parsing is commonly used when importing data from CSV files, Excel files, databases, APIs, and other external sources.

#4. Timedelta


`Timedelta` represents a difference between two dates or times.

It can be used to calculate how much time has passed between two dates or to add or subtract a specific amount of time.

## Demonstration

### Simple Example

Suppose an order was placed on August 20 and delivered on August 25.

The difference between these dates tells us how many days the delivery took.

### Business Example

An e-commerce company can calculate delivery time using the difference between order date and delivery date.

### AI/ML Example

Delivery time, customer waiting time, or time between events can be converted into numerical features for machine learning.

```python
import pandas as pd

order_date = pd.Timestamp("2026-08-20")
delivery_date = pd.Timestamp("2026-08-25")

difference = delivery_date - order_date

print(difference)
```

## Explanation

The two timestamps are subtracted from each other.

The result is a `Timedelta` showing the difference between the two dates.

## Where It Is Used

Timedelta is useful for calculating delivery time, employee working duration, customer waiting time, subscription duration, and time between events.

#5. Time Zones

A time zone represents the local time of a particular region of the world.

The same moment can have different local times in different countries.

Pandas can work with timezone-aware datetime values.

##  Demonstration

### Simple Example

Suppose a company has offices in India and the United States.

An event happening at the same moment will have different local times in these locations.

### Business Example

A global company needs time zones when managing meetings, orders, customer activity, and server logs across different countries.

### AI/ML Example

Time-zone information can be useful when analyzing user activity across different geographical locations.

```python
import pandas as pd

time = pd.Timestamp("2026-08-20 10:00:00", tz="Asia/Kolkata")

print(time)
```

## Explanation

The `tz` parameter assigns the `Asia/Kolkata` time zone to the timestamp.

This creates a timezone-aware timestamp.

## Where It Is Used

Time zones are important in global applications, international businesses, server logs, online meetings, financial systems, and user activity analysis.

# 6. Extracting Date Features


Date features are individual pieces of information extracted from a datetime value.

Common date features include:

- Year
- Month
- Day
- Day of Week
- Hour
- Minute
- Second

These features can help us understand patterns in time-based data.

## Demonstration

### Simple Example

Suppose we have a column containing order dates.

Instead of using the complete date, we can extract the year, month, day, or weekday.

### Business Example

A company can analyze which month has the highest sales or which day of the week receives the most orders.

### AI/ML Example

Year, month, day, weekday, and hour can be used as features in machine learning models to identify time-based patterns.

```python
import pandas as pd

df = pd.DataFrame({
    "Date": pd.to_datetime([
        "2026-08-20 10:30:00",
        "2026-09-15 14:45:00",
        "2026-10-10 18:20:00"
    ])
})

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Weekday"] = df["Date"].dt.day_name()
df["Hour"] = df["Date"].dt.hour

print(df)
```

## Explanation

The `.dt` accessor allows us to extract individual components from datetime values.

For example:

- `.dt.year` → extracts the year
- `.dt.month` → extracts the month
- `.dt.day` → extracts the day
- `.dt.day_name()` → extracts the weekday name
- `.dt.hour` → extracts the hour

## Where It Is Used

Extracting date features is commonly used in time-series analysis, sales forecasting, customer behavior analysis, demand prediction, and machine learning feature engineering.
