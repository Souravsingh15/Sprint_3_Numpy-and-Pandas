# 1. CSV Files


CSV stands for Comma-Separated Values.

A CSV file stores data in a simple table-like format. Each line usually represents one row, and commas separate the values in different columns.

CSV is one of the most common formats for sharing datasets because it is simple, lightweight, and supported by many different applications.

For example, a company could store employee information in a file called `employees.csv`.

Pandas provides:

- `read_csv()` for reading CSV files.
- `to_csv()` for writing DataFrames to CSV files.

## Demonstration

### Real-world example

Imagine an employee database containing thousands of employees.

The company could export the data as:

`employees.csv`

The file could contain information such as employee ID, name, department, and salary.

CSV is useful when data needs to be easily exchanged between different systems, spreadsheet applications, and programming languages.

### AI/ML Use Case

Many datasets used in machine learning are distributed as CSV files.

For example, a dataset downloaded from Kaggle may contain thousands of rows of training data. Pandas can load the CSV file into a DataFrame so that the data can be explored and prepared for a machine learning model.

```python
import pandas as pd
import numpy as np

data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Asha", "Rahul", "Priya", "Arjun", "Neha"],
    "Department": ["AI", "HR", "AI", "Finance", "Data Science"],
    "Salary": [50000, 42000, 58000, 45000, 62000]
}

df = pd.DataFrame(data)

print(df)
# Write the DataFrame to a CSV file

df.to_csv("employees.csv", index=False)

print("CSV file created successfully.")
```

```python
# Read the CSV file

df_csv = pd.read_csv("employees.csv")

print(df_csv)
```

### Explanation

`to_csv()` saves the DataFrame as a CSV file.

The argument `index=False` prevents Pandas from saving the DataFrame index as an additional column.

`read_csv()` reads the CSV file and converts it back into a Pandas DataFrame.

The expected output is the employee table that we originally created.

### Why use these methods?

- `to_csv()` is used when we want to save tabular data in CSV format.
- `read_csv()` is used when we want to load CSV data into Pandas.

### AI/ML Use

CSV files are commonly used for datasets in machine learning. A typical workflow can be:

CSV file → Pandas DataFrame → Data Cleaning → Feature Engineering → ML Model

# 3. JSON Files

JSON stands for JavaScript Object Notation.

JSON stores information in a structured format using key-value pairs and lists. It is commonly used when applications need to exchange data.

For example, information about an employee could be represented using keys such as:

- Employee_ID
- Name
- Department
- Salary

Pandas provides:

- `read_json()` to read JSON data.
- `to_json()` to write a DataFrame as JSON.

## Demonstration

### Real-world example

Suppose a web application sends customer information to a Python backend.

The information might be transferred in JSON format.

For example, an application could send employee details as structured JSON data.

JSON is especially common in web APIs because it can easily be exchanged between different programming languages and applications.

### AI/ML Use Case

AI applications often receive data through APIs. If an API returns information in JSON format, Pandas can be used to convert that data into a DataFrame for further analysis.

```python
# Write DataFrame to JSON

df.to_json("employees.json", orient="records", indent=4)

print("JSON file created successfully.")
```

```python
# Read JSON file

df_json = pd.read_json("employees.json")

print(df_json)
```

### Explanation

`to_json()` converts the DataFrame into JSON format and saves it to a file.

The `orient="records"` option stores each DataFrame row as a separate JSON object.

`read_json()` reads the JSON file and converts it into a DataFrame.

The expected output is the original employee table.

### Why use JSON?

JSON is useful when exchanging structured data between applications, especially through APIs.

### AI/ML Use

Many AI/ML applications consume data from APIs. If the API returns JSON data, Pandas can help convert it into a format suitable for analysis and preprocessing.

# 4. Parquet Files


Parquet is a file format designed for storing structured data efficiently.

Unlike CSV, which stores data mainly as plain text, Parquet stores data in a more organized and efficient form.

Parquet is particularly useful for large datasets because it supports efficient storage and reading of selected columns.

Pandas provides:

- `read_parquet()` for reading Parquet files.
- `to_parquet()` for writing Parquet files.

Parquet usually requires an additional library such as `pyarrow` or `fastparquet`.

## Demonstration

### Business example

Imagine a company has several gigabytes of customer transaction data.

Storing all of this information as CSV can require a lot of storage space and can make data processing slower.

Parquet is often a better choice for large analytical datasets because it is designed for efficient storage and processing.

### AI/ML Use Case

Large AI/ML datasets can contain millions of rows and many features. Parquet can be useful for storing these datasets efficiently, especially in data engineering and cloud-based analytics systems.

```python
# Write DataFrame to Parquet

df.to_parquet("employees.parquet", index=False)

print("Parquet file created successfully.")
```

```python
# Read Parquet file

df_parquet = pd.read_parquet("employees.parquet")

print(df_parquet)
```

### Explanation

`to_parquet()` saves the DataFrame in Parquet format.

`read_parquet()` reads the Parquet file back into a DataFrame.

The expected output is the employee dataset.

### Why use Parquet?

Parquet is useful when working with larger datasets where storage efficiency and faster analytical processing are important.

### AI/ML Use

Parquet is commonly used in data pipelines and large-scale data processing. Machine learning teams may store prepared datasets in Parquet before using them for model training or analysis.

# 5. Pickle Files

Pickle is a Python-specific format that can store Python objects in a file.

Instead of converting the data into a text-based format, Pickle can preserve Python objects so that they can be loaded back into Python later.

Pandas provides:

- `to_pickle()` for saving data.
- `read_pickle()` for loading data.

Pickle is convenient when the data will mainly be used within Python.

##  Demonstration

### Simple example

Suppose we have performed several operations on a Pandas DataFrame.

Instead of repeating those operations every time we start our program, we can save the DataFrame as a Pickle file and load it later.

### AI/ML Use Case

Pickle can be used to save Python objects used during data science workflows.

For example, a processed DataFrame can be saved and loaded later without repeating the same preprocessing steps.

However, Pickle files should only be loaded from trusted sources because loading an untrusted Pickle file can execute malicious code.

```python
# Save DataFrame using Pickle

df.to_pickle("employees.pkl")

print("Pickle file created successfully.")
```

```python
# Read Pickle file

df_pickle = pd.read_pickle("employees.pkl")

print(df_pickle)
```

### Explanation

`to_pickle()` stores the DataFrame as a Python Pickle file.

`read_pickle()` loads the stored object back into Python.

The expected output is the original DataFrame.

### Why use Pickle?

Pickle is useful when we want to preserve Python objects and use them again in a Python environment.

### AI/ML Use

Pickle can be useful for storing intermediate Python objects in a data science workflow. However, it should be used carefully because Pickle is Python-specific and untrusted Pickle files should never be loaded.

# 6. SQL Tables


SQL databases store structured data in tables consisting of rows and columns.

Unlike a CSV file stored on a computer, a database is designed to manage data efficiently and support operations such as searching, filtering, updating, and querying large amounts of information.

Pandas can work with SQL databases.

Two useful functions are:

- `pd.read_sql()` to read the result of a SQL query into a DataFrame.
- `DataFrame.to_sql()` to write a DataFrame into a SQL table.

For this example, I will use SQLite because it is lightweight and does not require a separate database server.

## Demonstration

### Business example

Imagine a company has millions of customer records.

Instead of storing everything in separate CSV files, the company can use a database.

An analyst can use SQL to request specific information, and Pandas can load the query result into a DataFrame for further analysis.

### AI/ML Use Case

Many companies store production data in SQL databases.

A typical ML workflow may be:

SQL Database → SQL Query → Pandas DataFrame → Data Preprocessing → ML Model

```python
import sqlite3

# Create a SQLite database
connection = sqlite3.connect("company.db")

# Write the DataFrame to a SQL table
df.to_sql("employees", connection, if_exists="replace", index=False)

print("Data successfully written to SQL table.")
```

```python
# Read data from the SQL table

df_sql = pd.read_sql("SELECT * FROM employees", connection)

print(df_sql)
```

```python
# Close the database connection

connection.close()
```

### Explanation

`sqlite3.connect()` creates a connection to a SQLite database.

`to_sql()` writes the Pandas DataFrame into a SQL table.

In this example, the table is called `employees`.

`pd.read_sql()` executes a SQL query and converts the result into a Pandas DataFrame.

The expected output is the employee information stored in the SQL table.

### Why use SQL?

SQL databases are useful when working with large, structured datasets that need to be stored, queried, and managed by multiple applications or users.

### AI/ML Use

Machine learning systems often obtain training or prediction data from databases. Pandas can act as a bridge between SQL databases and Python-based machine learning libraries.

# 7. Reading Multiple Files


Sometimes data is not stored in one large file. Instead, it may be divided into multiple files.

For example, a company may store monthly sales data as:

- `sales_january.csv`
- `sales_february.csv`
- `sales_march.csv`

If we need to analyze all three months together, we can read the files one by one and combine their contents.

Python's `glob` module can help find files that match a particular pattern.

## Demonstration

### Business example

A company may generate a new sales CSV file every month.

Instead of manually opening every file, a Python program can find all CSV files in a folder, read them, and combine them into one DataFrame.

### AI/ML Use Case

Machine learning datasets are sometimes divided into multiple files. For example, data collected from different days, locations, or machines may be stored separately.

Combining these files allows us to create a larger dataset for analysis or model training.

```python
import os

# Create a folder for the example files
os.makedirs("monthly_data", exist_ok=True)

# Create two sample DataFrames
january = pd.DataFrame({
    "Month": ["January", "January"],
    "Sales": [1000, 1200]
})

february = pd.DataFrame({
    "Month": ["February", "February"],
    "Sales": [1300, 1500]
})

# Save the files
january.to_csv("monthly_data/january.csv", index=False)
february.to_csv("monthly_data/february.csv", index=False)

print("Monthly files created.")
```

```python
import glob

files = glob.glob("monthly_data/*.csv")

print("Files found:")
print(files)
```

```python
dataframes = []

for file in files:
    temp_df = pd.read_csv(file)
    dataframes.append(temp_df)

combined_df = pd.concat(dataframes, ignore_index=True)

print(combined_df)
```

### Explanation

`glob.glob()` finds files that match a particular pattern.

In this example:

`monthly_data/*.csv`

means that we want to find CSV files inside the `monthly_data` folder.

Each file is read using `pd.read_csv()`.

The DataFrames are stored in a list, and `pd.concat()` combines them into one DataFrame.

The expected output contains the sales information from both January and February.

### AI/ML Use

This technique is useful when data is collected continuously and stored in separate files.

For example:

Day 1 data
→ Day 2 data
→ Day 3 data
→ combine
→ clean
→ train ML model

# 8. File Path Handling


A file path tells Python where a file is located.

For example:

`data/employees.csv`

means that the `employees.csv` file is inside the `data` folder.

Using file paths correctly is important because a program should be able to locate files reliably.

Python provides the `os` module and the `pathlib` module for working with file paths.

`pathlib` is especially convenient because it provides an object-oriented way to work with paths.

## Demonstration

### Real-world analogy

Think of a file path like a home address.

If someone knows only the name of a person but not their address, finding them is difficult.

Similarly, Python needs the correct path to locate a file.

For example:

`data/employees.csv`

tells Python that `employees.csv` is inside the `data` folder.

### AI/ML Use Case

ML projects often contain folders such as:

- `data/`
- `models/`
- `notebooks/`
- `results/`

Using proper file paths helps keep the project organized and makes the code easier to move between different machines.

```python
from pathlib import Path

# Create a path object
file_path = Path("employees.csv")

print("File name:", file_path.name)
print("File extension:", file_path.suffix)
print("File exists:", file_path.exists())
```

### Explanation

`Path()` creates a path object.

The `.name` attribute returns the file name.

The `.suffix` attribute returns the file extension.

The `.exists()` method checks whether the file actually exists at the specified location.

Using `pathlib` makes file path handling easier and more readable.

### Why use Path objects?

Hardcoding file paths as strings can become difficult when a project grows or when the code needs to run on different operating systems.

`pathlib` provides a more reliable way to construct and work with file paths.

### AI/ML Use

AI/ML projects often involve many datasets, models, configuration files, and output files. Proper path handling helps organize these resources and makes data pipelines easier to maintain.
