# 1. What is Pandas?

Pandas is a Python library used for working with structured data. It provides tools that make it easier to store, view, modify, filter, and analyze data.

One of the main advantages of Pandas is that it allows us to work with data in a table-like format, similar to an Excel spreadsheet or a database table.

Pandas mainly provides two important data structures:

- Series – represents one-dimensional data, similar to a single column.
- DataFrame – represents two-dimensional data, similar to a complete table.

Pandas is especially useful when working with datasets because we can perform many data manipulation operations with relatively simple Python commands.

# 2. Why Pandas?

Python already provides lists and dictionaries for storing data, so it is reasonable to ask why we need Pandas.

The main reason is that real-world datasets are often large and structured in rows and columns. Pandas provides convenient functions specifically designed for this kind of data.

With Pandas, we can easily:

- Work with rows and columns
- Filter data
- Sort data
- Find missing values
- Calculate statistics
- Add or remove columns
- Read data from files
- Save processed data to files

Instead of writing many lines of Python code to perform these operations manually, Pandas provides ready-to-use methods.

# Demonstration

### Business example

Suppose a company has information about its employees, including their department and salary.

The company may want to find:

- Employees with a salary above €40,000
- The average salary
- Employees belonging to a particular department
- The highest salary

Pandas can perform these operations directly on a DataFrame.

This is useful because business datasets can contain thousands or millions of records.

```python
import pandas as pd

employees = pd.DataFrame({
    "Name": ["Asha", "Rahul", "Priya", "Arjun"],
    "Department": ["AI", "HR", "AI", "Finance"],
    "Salary": [45000, 38000, 52000, 41000]
})

print(employees)
```

```python
high_salary = employees[employees["Salary"] > 40000]

print(high_salary)
```

### Explanation

The first code creates a DataFrame containing employee information.

The second code filters the DataFrame and keeps only employees whose salary is greater than 40,000.

The expression:

`employees["Salary"] > 40000`

checks every salary and produces a Boolean result.

Pandas then uses those Boolean values to select the required rows.

### AI/ML Use

Filtering and selecting data are important during data preprocessing. For example, we may remove invalid records or select observations that satisfy a particular condition before training a machine learning model.

# 3. Series

A Pandas Series is a one-dimensional data structure. It can be thought of as a single column of data.

Each value in a Series has an index. By default, Pandas creates an index starting from 0.

A Series can contain different types of values, such as numbers, text, or Boolean values.

For example, the marks of five students can be stored as a Series.

## Demonstration

### Real-world analogy

Think about one column in an Excel spreadsheet.

If the column contains only students' scores:

| Score |
|------:|
| 85 |
| 72 |
| 91 |
| 68 |
| 88 |

This single column can be represented using a Pandas Series.

```python
import pandas as pd

scores = pd.Series([85, 72, 91, 68, 88])

print(scores)
```

## Explanation

The `pd.Series()` function creates a Pandas Series.

In this example, the five numbers represent the scores of five students.

Pandas automatically creates an index:

- 0 for the first value
- 1 for the second value
- 2 for the third value
- and so on

The expected output contains two parts: the index on the left and the values on the right.

### AI/ML Use

A Series is useful when working with a single feature or column from a dataset. For example, a dataset may contain a Series representing age, salary, temperature, or exam scores.

# 4. DataFrame


A DataFrame is a two-dimensional Pandas data structure that organizes information into rows and columns.

It is similar to a table in a database or a spreadsheet in Excel.

Each column can contain a different type of information. For example, one column may contain names, another may contain ages, and another may contain numerical scores.

A DataFrame is one of the most important structures in Pandas because most real-world datasets are naturally represented as tables.

## Demonstration

### Real-world example

Suppose we have information about students:

| Name | Age | Python Score |
|------|-----|--------------|
| Asha | 21 | 85 |
| Rahul | 22 | 72 |
| Priya | 20 | 91 |

The entire table can be represented using a Pandas DataFrame.

Here:

- Each row represents one student.
- Each column represents one feature.
- The complete table represents the DataFrame.

```python
students = pd.DataFrame({
    "Name": ["Asha", "Rahul", "Priya"],
    "Age": [21, 22, 20],
    "Python_Score": [85, 72, 91]
})

print(students)
```

### Explanation

The `pd.DataFrame()` function creates a DataFrame.

In this example, the dictionary keys become column names, while the values become the data inside those columns.

The DataFrame contains three rows and three columns.

### AI/ML Use

Machine learning datasets are commonly represented as DataFrames.

For example:

- Rows can represent individual customers.
- Columns can represent customer features.
- One column can represent the target variable.

Pandas can then be used to prepare this DataFrame before giving the data to a machine learning algorithm.

# 5. Creating DataFrames


A DataFrame can be created in several ways depending on where the data comes from.

Common methods include creating a DataFrame from:

- A dictionary
- A list of dictionaries
- A NumPy array
- Existing data

The most convenient method for simple structured data is often a dictionary, where the keys represent column names and the values represent the column data.

```python
import numpy as np
import pandas as pd

data = {
    "Name": ["Asha", "Rahul", "Priya"],
    "Age": [21, 22, 20],
    "Score": [85, 72, 91]
}

df_dict = pd.DataFrame(data)

print(df_dict)
```

### Creating a DataFrame from a NumPy array

NumPy arrays can also be converted into DataFrames.

This is useful when numerical data has already been stored or processed using NumPy.

```python
data_array = np.array([
    [21, 85],
    [22, 72],
    [20, 91]
])

df_numpy = pd.DataFrame(
    data_array,
    columns=["Age", "Score"]
)

print(df_numpy)
```

### Explanation

`pd.DataFrame()` converts structured data into a table-like DataFrame.

When using a NumPy array, the `columns` parameter gives meaningful names to the columns.

### AI/ML Use

DataFrames are often created during data preprocessing. For example, numerical data generated by NumPy or returned by another process can be converted into a DataFrame so that Pandas operations can be performed on it.

# 6. Data Types

A data type tells us what kind of value is stored in a column.

For example:

- Integer values are whole numbers.
- Float values contain decimal numbers.
- String values contain text.
- Boolean values contain True or False.

Knowing the data types is important because different operations can be performed on different types of data.

For example, we can calculate the average of a numerical column, but calculating the average of a column containing names does not make sense.

```python
data = {
    "Name": ["Asha", "Rahul", "Priya"],
    "Age": [21, 22, 20],
    "Score": [85.5, 72.0, 91.5],
    "Passed": [True, True, True]
}

df = pd.DataFrame(data)

print(df.dtypes)
```

### Explanation

The `dtypes` attribute shows the data type of each column.

In this example:

- `Name` contains text.
- `Age` contains integer values.
- `Score` contains decimal values.
- `Passed` contains Boolean values.

Checking data types is an important part of understanding a dataset.

### AI/ML Use

Machine learning algorithms generally require numerical input. Therefore, we often need to identify text or categorical columns and convert them into numerical representations during preprocessing.

# 7. Index

An index is used to identify the rows of a Pandas Series or DataFrame.

By default, Pandas assigns numbers starting from 0.

However, we can also create our own index values.

An index makes it easier to identify and access specific rows.

```python
students = pd.DataFrame({
    "Name": ["Asha", "Rahul", "Priya"],
    "Score": [85, 72, 91]
})

print(students.index)
```

```python
students.index = ["Student_A", "Student_B", "Student_C"]

print(students)
```

# 8. Columns

Columns represent different variables or features in a DataFrame.

For example, in a student dataset:

- `Name` is one column.
- `Age` is another column.
- `Score` is another column.

We can access, modify, add, rename, or remove columns using Pandas.

```python
students = pd.DataFrame({
    "Name": ["Asha", "Rahul", "Priya"],
    "Age": [21, 22, 20],
    "Score": [85, 72, 91]
})

print(students.columns)
```

```python
print(students["Score"])
```

```python
students["Passed"] = students["Score"] >= 50

print(students)
```

### Explanation

`df.columns` returns the names of all columns in the DataFrame.

Using:

`students["Score"]`

allows us to select one specific column.

We can also create a new column by assigning values to a new column name.

In this example, the `Passed` column is created by checking whether each student's score is greater than or equal to 50.

### AI/ML Use

Columns often represent features used by machine learning models.

For example, in a house price prediction dataset, columns might include:

- Area
- Number of bedrooms
- Location
- Age of the house

These columns can be selected and processed before training the model.

# 9. Reading Data

Real-world data is usually stored in files rather than directly inside a Python program.

Pandas provides functions that allow us to read data from different file formats.

One of the most common formats is CSV (Comma-Separated Values).

The `pd.read_csv()` function is used to load a CSV file into a DataFrame.

```python
# Create a small CSV file first

students.to_csv("students.csv", index=False)

print("CSV file created successfully.")
```

```python
df = pd.read_csv("students.csv")

print(df)
```

### Explanation

`to_csv()` saves a DataFrame as a CSV file.

`pd.read_csv()` reads a CSV file and converts its contents into a Pandas DataFrame.

The `index=False` argument prevents Pandas from saving the DataFrame index as an additional column in the CSV file.

### AI/ML Use

Reading data is usually one of the first steps in an AI/ML project.

For example, a dataset downloaded from Kaggle or provided by a company may be stored as a CSV file. Pandas can load this dataset into a DataFrame so that it can be explored and prepared for machine learning.

# 10. Writing Data

Writing data means saving a DataFrame into a file so that it can be used later or shared with another person or application.

Pandas provides different methods for saving data. One common method is `to_csv()`.

This is useful because data that has been cleaned or transformed can be saved instead of performing the same processing again.

```python
students.to_csv("students_output.csv", index=False)

print("Data successfully saved to students_output.csv")
```

### Explanation

The `to_csv()` method writes the DataFrame to a CSV file.

The file `students_output.csv` will contain the data from the DataFrame.

The `index=False` argument means that the Pandas index will not be written into the CSV file.

### AI/ML Use

Writing data is useful after data preprocessing.

For example, we may:

1. Load a raw dataset.
2. Clean missing values.
3. Remove unnecessary columns.
4. Transform features.
5. Save the cleaned dataset.

The cleaned dataset can then be used for machine learning or shared with other members of a data science team.
