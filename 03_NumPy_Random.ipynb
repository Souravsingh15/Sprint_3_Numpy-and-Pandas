# 1. Random Numbers

Random numbers are values generated in a way that does not follow a predictable sequence from the user's point of view.

NumPy provides the `np.random` module for generating random numerical values.

Random numbers are useful when we need simulated or randomly generated data.

### Example

Suppose we want to simulate the result of rolling a virtual experiment many times.

Instead of manually entering the results, we can generate random values using NumPy.

### AI/ML Use Case

Random numbers are used in simulations, model initialization, data sampling, and creating synthetic datasets.

```python
import numpy as np

random_numbers = np.random.rand(5)

print(random_numbers)
```

### Explanation

`np.random.rand(5)` generates 5 random floating-point numbers between 0 and 1.

The exact output will be different each time the code runs unless a random seed is set.

# 2. Random Seed

A random seed allows us to control the starting point of random number generation.

Using the same seed produces the same sequence of random numbers.

This is useful when we want our experiments to be reproducible.

### Example

If two researchers use the same random seed, they can generate the same random values.

### AI/ML Use Case

Random seeds are commonly used in machine learning experiments so that results can be reproduced and compared fairly.

```python
import numpy as np

np.random.seed(42)

numbers = np.random.rand(5)

print(numbers)
```

### Explanation

`np.random.seed(42)` sets the starting point for the random number generator.

Running the code again with the same seed produces the same 5 values.

# 3. Random Integers

Random integers are randomly generated whole numbers.

NumPy provides `np.random.randint()` for generating random integers within a specified range.

### Example

If we want to simulate rolling a six-sided die, we can generate random integers from 1 to 6.

### AI/ML Use Case

Random integers can be used for sampling, simulations, creating random IDs, and generating test data.

```python
import numpy as np

dice_rolls = np.random.randint(1, 7, size=10)

print("Dice rolls:", dice_rolls)
```

### Explanation

`np.random.randint(1, 7, size=10)` generates 10 random integers from 1 up to, but not including, 7.

Therefore, the possible values are 1 through 6.

# 4. Random Choice

Random choice selects random elements from an existing array or collection.

NumPy provides `np.random.choice()` for this purpose.

### Example

Suppose we have:

["Red", "Blue", "Green"]

We can randomly select one of these options.

### Business Example

A company could randomly select customers for a survey.

### AI/ML Use Case

Random choice can be used for sampling observations from a dataset.

```python
import numpy as np

customers = np.array(["Customer A", "Customer B", "Customer C", "Customer D"])

selected = np.random.choice(customers)

print("Selected customer:", selected)
```

### Explanation

`np.random.choice()` randomly selects one element from the array.

The selected customer can be different each time the code runs.

# 5. Shuffle

Shuffling randomly changes the order of elements in an array.

NumPy provides `np.random.shuffle()` for shuffling an array in place.

### Example

Original data:

[1, 2, 3, 4, 5]

After shuffling, the order might become:

[3, 1, 5, 2, 4]

### AI/ML Use Case

Shuffling data is useful before training machine learning models so that the original ordering of observations does not influence the training process.

```python
import numpy as np

data = np.array([1, 2, 3, 4, 5])

np.random.shuffle(data)

print("Shuffled data:", data)
```

### Explanation

`np.random.shuffle()` randomly rearranges the elements of the existing array.

The original array itself is changed.

# 6. Permutation

A permutation produces a randomly reordered version of a sequence.

Unlike `shuffle()`, `np.random.permutation()` returns a new array and does not modify the original array.

### Example

Original:

[1, 2, 3, 4, 5]

A random permutation could be:

[4, 1, 5, 2, 3]

### AI/ML Use Case

Random permutations can be useful when randomly rearranging observations or performing experiments involving different data orders.

```python
import numpy as np

data = np.array([1, 2, 3, 4, 5])

permuted = np.random.permutation(data)

print("Original:", data)
print("Permutation:", permuted)
```

### Explanation

`np.random.permutation()` creates a randomly reordered copy of the array.

The original `data` remains unchanged.

# 7. Random Distributions

A random distribution describes how randomly generated values are spread across possible outcomes.

NumPy can generate random values from different probability distributions.

Examples include:

- Normal distribution
- Uniform distribution
- Binomial distribution
- Poisson distribution

### Example

A normal distribution can be used to simulate values that tend to cluster around an average.

For example, we can simulate heights of people using a normal distribution.

### AI/ML Use Case

Random distributions are used to create simulated data, initialize models, perform statistical experiments, and test machine learning algorithms.

```python
import numpy as np

normal_data = np.random.normal(
    loc=170,
    scale=10,
    size=1000
)

print("First 10 values:")
print(normal_data[:10])
```

### Explanation

`np.random.normal()` generates random values following a normal distribution.

`loc=170` represents the mean.

`scale=10` represents the standard deviation.

`size=1000` generates 1000 values.

The generated values will generally be concentrated around 170.
