# 1. concat()

`concat()` is used to combine two or more Pandas DataFrames.

It can combine DataFrames row-wise or column-wise.

When combining row-wise, one DataFrame is placed below another DataFrame.

## Demonstration

### Real-world Example

Suppose a school has student records for Class A and Class B in two different DataFrames.

If we want to create one DataFrame containing students from both classes, we can use `concat()`.

### Business Example

A company may have sales data for January and February stored separately. We can use `concat()` to combine both datasets.

### AI/ML Example

In machine learning, data may be collected in different batches. `concat()` can combine these batches into one dataset before preprocessing and model training.

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Sourav", "Rahul"],
    "Marks": [85, 90]
})

df2 = pd.DataFrame({
    "Name": ["Amit", "Ravi"],
    "Marks": [78, 88]
})

result = pd.concat([df1, df2], ignore_index=True)

print(result)
```

## Explanation

The code creates two DataFrames and combines them row-wise using `pd.concat()`.

`ignore_index=True` creates a new continuous index for the combined DataFrame.

## Where It Is Used

`concat()` is commonly used when data is divided into multiple files, batches, months, or departments and needs to be combined into one dataset.

#2. merge()


`merge()` is used to combine two DataFrames using one or more common columns.

It is similar to a JOIN operation in a database.

For example, if two DataFrames have a common `student_id`, we can use that column to combine them.

##  Demonstration

### Real-world Example

Suppose one DataFrame contains student names and another contains student marks.

Both DataFrames have `student_id`.

We can use `merge()` to combine the student's name and marks.

### Business Example

A company may have customer information in one dataset and customer transaction information in another dataset. Both datasets can be combined using `customer_id`.

### AI/ML Example

Customer information and transaction data can be merged before creating features for a machine learning model.

```python
students = pd.DataFrame({
    "student_id": [1, 2, 3],
    "name": ["Sourav", "Rahul", "Amit"]
})

marks = pd.DataFrame({
    "student_id": [1, 2, 3],
    "marks": [85, 90, 78]
})

result = pd.merge(students, marks, on="student_id")

print(result)
```

## Explanation

`student_id` is the common column between the two DataFrames.

`merge()` matches the same `student_id` values and combines the corresponding information.

## Where It Is Used

`merge()` is useful when related information is stored in different datasets and the datasets have a common key such as customer_id, product_id, or student_id.

#3. join()


`join()` is used to combine DataFrames mainly using their index.

It is useful when two DataFrames already have a common index.

## Demonstration

### Real-world Example

Suppose one DataFrame contains student names and another contains their marks.

If both DataFrames use the same student ID as their index, we can use `join()` to combine them.

### Business Example

A company may store employee details and salary information in separate DataFrames using employee ID as the index.

### AI/ML Example

Different feature DataFrames may use the same index for observations. `join()` can combine those features into one DataFrame.

```python
students = pd.DataFrame({
    "Name": ["Sourav", "Rahul", "Amit"]
}, index=[1, 2, 3])

marks = pd.DataFrame({
    "Marks": [85, 90, 78]
}, index=[1, 2, 3])

result = students.join(marks)

print(result)
```

## Explanation

Both DataFrames have the same index values: 1, 2, and 3.

`join()` matches these indexes and combines the columns.

## Where It Is Used

`join()` is useful when related DataFrames already share a common index and we want to combine their columns.

#4. append()

`append()` was an older Pandas method used to add the rows of one DataFrame to another DataFrame.

However, `DataFrame.append()` was deprecated in Pandas 1.4.0 and removed in Pandas 2.0.

So, we should not use `append()` in new code.

The recommended alternative is `pd.concat()`.

## Why was append() deprecated?

`append()` was mainly used for combining DataFrames, and `concat()` provides the same functionality in a more flexible and consistent way.

Instead of:

df1.append(df2)

we should use:

pd.concat([df1, df2])

## Demonstration

### Real-world Example

Suppose we have an existing student dataset and receive new student records.

We want to add the new records to the existing DataFrame.

Earlier, `append()` was commonly used for this purpose.

Today, we should use `pd.concat()` instead.

### Business Example

A company receives new sales records and needs to add them to an existing sales dataset.

### AI/ML Example

New batches of training data can be added to an existing dataset using `pd.concat()`.

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Sourav", "Rahul"],
    "Marks": [85, 90]
})

df2 = pd.DataFrame({
    "Name": ["Amit", "Ravi"],
    "Marks": [78, 88]
})

result = pd.concat([df1, df2], ignore_index=True)

print(result)
```

## Explanation

The code combines the rows of `df1` and `df2` using `pd.concat()`.

`ignore_index=True` creates a new continuous index after combining the DataFrames.

This is the modern alternative to the deprecated `append()` method.

## Where It Is Used

`pd.concat()` can be used when new rows or datasets need to be added to an existing DataFrame.

It should be preferred over the deprecated `append()` method.


#5. combine_first()


`combine_first()` is used to fill missing values in one DataFrame using values from another DataFrame.

If the first DataFrame already contains a value, that value is kept.

If the first DataFrame contains a missing value, the value from the second DataFrame is used.

## Demonstration

### Real-world Example

Suppose we have student marks from two different sources.

Some marks are missing in the first dataset, but those marks are available in the second dataset.

We can use `combine_first()` to fill the missing values.

### Business Example

A company may collect customer information from two different sources. If some information is missing in the main dataset, another dataset can provide the missing values.

### AI/ML Example

During data preprocessing, information from multiple sources can be combined to reduce missing values before training a machine learning model.

```python
df1 = pd.DataFrame({
    "Name": ["Sourav", "Rahul", "Amit"],
    "Marks": [85, None, 78]
})

df2 = pd.DataFrame({
    "Name": ["Sourav", "Rahul", "Amit"],
    "Marks": [90, 88, 80]
})

result = df1.combine_first(df2)

print(result)
```

## Explanation

`df1` has a missing value for Rahul's marks.

`combine_first()` keeps the values already available in `df1`.

Where `df1` has a missing value, it takes the corresponding value from `df2`.

## Where It Is Used

`combine_first()` is useful when we have multiple sources of data and want to fill missing values using another DataFrame.
