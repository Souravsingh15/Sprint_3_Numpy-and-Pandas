## 1. isnull()

`isnull()` is used to identify missing values in a Pandas DataFrame or Series.

It returns `True` where a value is missing and `False` where a value is present.

### Demonstration

Suppose a student dataset contains some students whose age or marks are not available.

`isnull()` helps us find which values are missing.

For example:

- Age = 20 → False
- Age = NaN → True

This is useful before deciding how to handle missing data.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Sourav"],
    "Age": [20, np.nan, 22],
    "Marks": [85, 90, np.nan]
})

df.isnull()
```

### Explanation

The code checks every value in the DataFrame for missing data.

`True` means the value is missing, while `False` means the value is present.

This helps us identify missing values before performing data analysis or machine learning.

## 2. notnull()

`notnull()` is used to identify values that are present in a Pandas DataFrame or Series.

It returns `True` when a value is present and `False` when the value is missing.

### Demonstration

Suppose we want to find students whose marks are available.

`notnull()` can identify the rows where marks are present.

For example:

- Marks = 85 → True
- Marks = NaN → False

```python
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Sourav"],
    "Marks": [85, np.nan, 90]
})

df["Marks"].notnull()
```

### Explanation

The code checks which values in the `Marks` column are available.

`True` represents a non-missing value and `False` represents a missing value.

It can also be used to filter rows that contain valid data.

## 3. dropna()

`dropna()` is used to remove rows or columns containing missing values.

It is useful when missing values cannot be reasonably filled or when incomplete records are not required.

### Demonstration

Suppose a student dataset contains three students, but one student has no marks.

If the missing record is not useful for our analysis, we can remove that row using `dropna()`.

```python
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Sourav"],
    "Marks": [85, np.nan, 90]
})

df.dropna()
```

### Explanation

The code removes the row containing the missing value.

The result contains only rows where the required values are available.

`dropna()` is useful when removing incomplete records is better than filling the missing values.

## 4. fillna()

`fillna()` is used to replace missing values with a specified value or calculated value.

Instead of deleting incomplete records, we can use `fillna()` to keep them.

### Demonstration

Suppose the marks of one student are missing.

Instead of deleting that student, we can replace the missing marks with the average marks of the other students.

This allows us to keep the record in the dataset.

```python
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Sourav"],
    "Marks": [85, np.nan, 90]
})

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

df
```

### Explanation

The code calculates the mean of the available marks and uses it to replace the missing value.

`fillna()` is useful when we want to preserve rows instead of removing them.

In machine learning, missing numerical values are often filled using methods such as mean or median imputation.

## 5. interpolate()

`interpolate()` estimates missing values using the values around them.

It is especially useful for ordered or time-series data where the missing value can be estimated from nearby observations.

### Demonstration

Suppose temperature measurements are:

10, 12, NaN, 16

The missing value can be estimated from the surrounding values.

The value between 12 and 16 is 14.

So interpolation can fill the missing value as 14.

```python
df = pd.DataFrame({
    "Temperature": [10, 12, np.nan, 16, 18]
})

df["Temperature"] = df["Temperature"].interpolate()

df
```

### Explanation

The code uses the surrounding values to estimate the missing temperature.

The missing value between 12 and 16 is filled with 14.

`interpolate()` is particularly useful for time-series and sequential data where values have a meaningful order.

## 6. Missing Value Handling Strategies

Missing values should be handled based on the type of data and the reason why values are missing.

Common strategies include:

- Remove rows using `dropna()`
- Remove columns when too many values are missing
- Fill numerical values using mean or median
- Fill categorical values using mode
- Use `interpolate()` for ordered or time-series data
- Use forward fill or backward fill when appropriate

The goal is to handle missing data without introducing unnecessary bias into the dataset.

### Demonstration

Suppose a dataset contains a `Salary` column with a few missing values.

We could use the median salary to fill the missing values if the salary data contains extreme values.

For time-series data such as daily temperature, interpolation may be more appropriate.

Therefore, the best strategy depends on the type and distribution of the data.

```python
df = pd.DataFrame({
    "Age": [20, 21, np.nan, 24, 25],
    "Salary": [30000, 35000, np.nan, 40000, 45000]
})

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

df
```

### Explanation

The code replaces missing values in the `Age` and `Salary` columns using their respective median values.

Median is useful when numerical data may contain extreme values because it is less affected by outliers than the mean.

The appropriate missing-value strategy should always depend on the dataset and the purpose of the analysis.
