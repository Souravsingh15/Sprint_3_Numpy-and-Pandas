#Notebook 12 – String Operations

## 1. lower()

The `lower()` method converts uppercase letters in a string into lowercase letters.

It is useful when we want text to have a consistent format.

### Demonstration

Suppose a dataset contains names like `"RAJU"`, `"Raju"`, and `"raju"`.

Converting them to lowercase makes them consistent:

`"raju"`, `"raju"`, `"raju"`

This is useful during data cleaning and text preprocessing.

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["RAJU", "Raju", "SOURAV"]
})

df["name"] = df["name"].str.lower()

df
```

### Explanation

The code converts every name in the `name` column into lowercase using `str.lower()`.

This helps maintain consistency in text data and is commonly used during data cleaning and AI/ML text preprocessing.

## 2. upper()

The `upper()` method converts lowercase letters in a string into uppercase letters.

It is useful when text needs to follow a common uppercase format.

### Demonstration

Suppose product categories are stored as:

`"laptop"`, `"mobile"`, `"tablet"`

We can convert them into:

`"LAPTOP"`, `"MOBILE"`, `"TABLET"`

This makes the text format consistent.

```python
df = pd.DataFrame({
    "category": ["laptop", "mobile", "tablet"]
})

df["category"] = df["category"].str.upper()

df
```

### Explanation

The code converts all values in the `category` column to uppercase.

`str.upper()` is useful for standardizing text values during data cleaning and analysis.

## 3. strip()

The `strip()` method removes unnecessary spaces from the beginning and end of a string.

It does not remove spaces between words.

### Demonstration

Suppose customer names are stored as:

`"  Rahul  "`

The extra spaces can cause problems when comparing names.

Using `strip()` converts it to:

`"Rahul"`

```python
df = pd.DataFrame({
    "name": ["  Rahul  ", " Priya", "Sourav "]
})

df["name"] = df["name"].str.strip()

df
```

### Explanation

The code removes leading and trailing spaces from every value in the `name` column.

`strip()` is useful when cleaning data collected from forms, files, or user input.

## 4. split()

The `split()` method breaks a string into smaller parts based on a separator.

By default, it separates text using whitespace.

### Demonstration

Suppose we have:

`"Python is easy"`

Using `split()` gives:

`["Python", "is", "easy"]`

This allows us to work with individual words separately.

```python
df = pd.DataFrame({
    "sentence": [
        "Python is easy",
        "Pandas is useful"
    ]
})

df["words"] = df["sentence"].str.split()

df
```

### Explanation

The code splits each sentence into individual words and stores them as a list.

`split()` is useful in basic text processing and can be used before further text analysis.

## 5. contains()

The `contains()` method checks whether a particular text or pattern exists inside a string.

It returns a Boolean value such as `True` or `False`.

### Demonstration

Suppose we have customer feedback:

`"The delivery was fast"`

We can check whether the feedback contains the word `"delivery"`.

The result will be `True`.

This can be useful for filtering rows containing specific words.

```python
df = pd.DataFrame({
    "feedback": [
        "The delivery was fast",
        "The product was damaged",
        "Fast delivery and good service"
    ]
})

df["has_delivery"] = df["feedback"].str.contains("delivery")

df
```

### Explanation

The code checks whether the word `"delivery"` appears in each feedback value.

The result is `True` when the word is present and `False` when it is not.

This is useful for filtering customer feedback and performing text-based analysis.

## 6. replace()

The `replace()` method changes an existing value or piece of text into another value.

It is useful for correcting or standardizing data.

### Demonstration

Suppose a city column contains:

`"Calcutta"`

but we want to standardize it as:

`"Kolkata"`

We can use `replace()` to change the value.

```python
df = pd.DataFrame({
    "city": ["Calcutta", "Delhi", "Calcutta", "Mumbai"]
})

df["city"] = df["city"].str.replace("Calcutta", "Kolkata")

df
```

### Explanation

The code searches for `"Calcutta"` in the `city` column and replaces it with `"Kolkata"`.

`replace()` is useful for correcting inconsistent text and cleaning datasets before analysis or machine learning.

## 7. Regular Expressions (Regex)

Regular Expressions, commonly called Regex, are patterns used to search, identify, or extract specific types of text.

Instead of searching for only one exact value, Regex allows us to define a pattern.

For example, a pattern can be used to identify email addresses or phone numbers.

### Demonstration

Suppose a dataset contains text such as:

`"Contact us at support@gmail.com"`

We can use a Regex pattern to identify the email address from the text.

Regex is useful when the information follows a particular pattern.

```python
import pandas as pd

df = pd.DataFrame({
    "text": [
        "Contact: rahul@gmail.com",
        "Contact: priya@yahoo.com",
        "No email available"
    ]
})

df["email"] = df["text"].str.extract(
    r'([A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,})'
)

df
```

### Explanation

The code uses a Regex pattern to search for email addresses inside the `text` column.

`str.extract()` finds the matching pattern and extracts it into a new column.

Regex is useful for data cleaning, information extraction, and text preprocessing in AI/ML applications.
